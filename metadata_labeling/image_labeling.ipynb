```python
import os
import csv
from PIL import Image
import pandas as pd
from IPython.display import display, clear_output
```

```python
## DO NOT CHANGE THIS ## 
input_id_field_map = {
    0: 'Site',
    1: 'Day',
    2: 'Plant',
    3: 'Level',
    4: 'Index'
}
```

```python
# Directory containing the images
## CHANGE THIS ##
image_dir = '/fs/ess/PAS2136/Hawaii-2025/invasive-plants/images/png/session_9_1_27_2025'

# CSV file to save inputs
## CHANGE THIS ##
csv_file = 'SITE_9.csv'
```

```python
# Display image and get metadata
def display_image_and_get_metadata(image_path):
    image = Image.open(image_path)
    image.thumbnail((600, 400))

    # Display the image inline in the Jupyter notebook
    display(image)
    
    # Get user inputs
    inputs = []
    for i in range(5):
        user_input = input(f"Enter {input_id_field_map[i]} for {os.path.basename(image_path)}: ")
        inputs.append(user_input)
    
    return inputs

# Write headers if new file, or read existing file to skip images already labeled 
already_labeled_files = None
if not os.path.exists(csv_file):
    with open(csv_file, 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Filename', 'Site', 'Day', 'Plant', 'Level', 'Index'])  # Header
else:
    already_labeled_files = list(pd.read_csv(csv_file).Filename)

required_files = os.listdir(image_dir)
total_count = len(required_files)
total_labeled = 0

if already_labeled_files:
    required_files = [f for f in required_files if f not in already_labeled_files]
    total_labeled = len(already_labeled_files)


# Iterate through each image in the directory
for filename in required_files:
    if filename.endswith('.jpg') or filename.endswith('.png'): 
        image_path = os.path.join(image_dir, filename)
        
        # Clear the previous image
        clear_output(wait=True)
        

        print(f'Image {total_labeled+1}/{total_count}:')
        # Get inputs for the current image
        inputs = display_image_and_get_metadata(image_path)
        
        # Save the inputs to the CSV
        with open(csv_file, 'a', newline='') as file:
            writer = csv.writer(file)
            writer.writerow([filename] + inputs)

        total_labeled += 1
```
